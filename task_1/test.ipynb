```python
import pandas as pd

cleaning_data = pd.read_csv('cleaning_data.csv')

dates = pd.to_datetime(
    cleaning_data['date'],
    format='mixed',
    errors='coerce'
)

# Display rows with invalid dates
print(cleaning_data[dates.isna()])
```

```python
cleaning_data =cleaning_data.sort_values(by='date')

print(cleaning_data)
```

```python
import numpy as np

cleaning_data['date'] = cleaning_data['date'].replace(['0', 0], np.nan)
```

```python
new_cleaning_data = cleaning_data.dropna(subset=['date'])
print(new_cleaning_data)
```

```python
new_cleaning_data.to_csv('cleaning_data.csv', index=False)
```
